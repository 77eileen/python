### K-means k-평균-비지도학습-군집(cluster)
- step1. 초기 중심점 설정(centroid)
- step2. 데이터할당(Assignment)
- step3. 중심점 업데이트 (Update)

```python
#6명의 학생 수학 영어 점수 2개의 그룹으로 나눈다
import matplotlib.pyplot as plt
import numpy as np
X = np.array ([
        [1,1], [1,2], [2,1],
        [5,4], [5,5], [6,5]
])
plt.scatter(X[:, 0], X[:,1], s=100, color='grey')
plt.grid(True)
plt.legend()
plt.show()
```

```python
#초기 중심점 할당
centroid_initial = np.array([[1,1], [5,4]])

plt.scatter(X[:, 0], X[:,1], s=100, color='grey', label='Data Point')
plt.scatter(centroid_initial[:, 0], centroid_initial[:,1], s=100, color=['red', 'blue'], marker='X', label='initial Point')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# 데이터 할당 : 모든 데이터 포인트를 순회하면서 두 중심점 중에서 더 가까운 클러스터로 할당
# 빨강클러스터 p1 p2 p3
# 파랑클러스터 p4 p5 p6

# 각 점이 어느 중심점에 더 가까운지 계산
label_inital = np.array([0,0,0,1,1,1])
plt.scatter(X[label_inital==0,0], X[label_inital==0,1], s=100, color='lightcoral', label='clusters1' )
plt.scatter(X[label_inital==1,0], X[label_inital==1,1], s=100, color='purple', label='clusters2' )
plt.scatter(centroid_initial[:, 0], centroid_initial[:,1], s=100, color=['red', 'blue'], marker='X', label='initial Point')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# 중심점 업데이트
c1_mean = np.mean(X[label_inital==0],axis=0)
c2_mean = np.mean(X[label_inital==1],axis=0)
centroid_updated = np.array([c1_mean, c2_mean])
plt.scatter(X[label_inital==0,0], X[label_inital==0,1], s=100, color='lightcoral', label='clusters1' )
plt.scatter(X[label_inital==1,0], X[label_inital==1,1], s=100, color='purple', label='clusters2' )
plt.scatter(centroid_initial[:, 0], centroid_initial[:,1], s=100, color=['red', 'blue'], marker='X', label='old centroid')
plt.scatter(centroid_updated[:, 0], centroid_updated[:, 1], s=100, color=['black', 'orange'], marker='^', label='new centroid')

plt.grid(True)
plt.legend()
plt.show()
```

```python
from sklearn.cluster import KMeans
kms = KMeans(n_clusters=2)
kms.fit(X)
```

```python
kms.labels_
#출력 해석 : array([0, 0, 0, 1, 1, 1], dtype=int32)
#각 좌표의 클러스터를 확인할 수 있음.
```

```python
from sklearn.cluster import KMeans
kms = KMeans(n_clusters=2)

X = np.array ([
        [5,4], [1,2], [5,5],[1,1],[2,1], [6,5]
])

kms.fit(X)

kms.labels_

#상기와 같이 X좌표 순서를 바꿔도 클러스터 알려줌을 확인
```

```python
kms.cluster_centers_  #클러스터의 중심점을 알려줌
```

```python
plt.scatter(X[:,0], X[:,1], color='grey')
plt.scatter(kms.cluster_centers_[:,0],kms.cluster_centers_[:,1],color='red')
plt.show()
# 앞에 우리가 계산한 것과 동일하게 중심점이 나옴
```
