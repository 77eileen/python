#### 🔹 SMOTE, SMOTEENN

- 데이터가 많고 깨끗하면 → SMOTE만
- 데이터가 noisy(헷갈리는 샘플 많음)하면 → SMOTEENN


1️⃣ SMOTE (소수 클래스 늘리기)

- 상황 예시:

    반 친구들 중에서 남자 2명, 여자 8명이 있어요.

    남자 친구 데이터를 더 늘리고 싶은데, 그냥 똑같이 복사하면 똑같은 친구가 계속 생기는 것처럼 모델이 헷갈려요.

- SMOTE 방식:

    남자 친구 2명 사이를 이어서 새로운 가상의 남자 친구를 만들어내요.

    예를 들어, 두 남자 친구 키가 160cm, 170cm면, 165cm 정도인 새 친구를 만들어서 남자 데이터를 늘리는 느낌이에요.

- 정리:
    - 그냥 복사 X
    - 두 샘플 사이에 새로운 가짜 데이터 생성
    - 데이터가 다양해져서 모델이 덜 헷갈림


2️⃣ SMOTEENN (SMOTE + 노이즈 제거)

- 상황 예시:

    SMOTE로 남자 친구를 늘렸는데, 주변에 여자 친구와 헷갈릴 수 있는 애들이 섞여 있어요.

    예를 들어, 키 160cm인데 옷 스타일이 여자 친구랑 비슷한 애가 있을 수 있어요.

- SMOTEENN 방식:

    먼저 SMOTE처럼 가짜 남자 친구를 만들어요.

    그 다음, 주변 친구들을 확인해서 **잘못 섞인 친구(노이즈)**를 삭제해요.

    즉, 남자 친구 데이터가 깨끗해지는 것

- 정리:

    - SMOTE로 늘린 후 주변 헷갈리는 애들 제거
    - 더 깨끗하고 모델이 잘 배울 수 있는 데이터가 됨

3️⃣ 한 줄 요약

- SMOTE: 남자 친구 수를 늘림 → 다양하지만 조금 헷갈릴 수 있음
- SMOTEENN: 남자 친구 수 늘리고 → 주변 헷갈리는 애들은 삭제 → 더 깨끗함
