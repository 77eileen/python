```python
# Category 컬럼 중 TREA의 데이터 갯수가 6으로 작아서 데이터 삭제
vc = df.Category.value_counts() 
keep = vc[vc > 6].index
df_filtered = df[df.Category.isin(keep)]
```

```python
# 데이터 샘플링 (20% 만 가져옴??)
from sklearn.model_selection import train_test_split
X_ = df_filtered.loc[:, 'X':]
y_ = df_filtered.Category
X,_, y,_ = train_test_split(X_,y_,stratify=y_, train_size=0.2, random_state=42)
X.shape
```

```python
# 8:2
from sklearn.svm import LinearSVC
svm =LinearSVC()
X_train, X_test, y_train, y_test = train_test_split(X,y,stratify=y,test_size=0.3,random_state=42)


svm.fit(X_train, y_train)
y_pred = svm.predict(X_test)

from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))

# 출력 결과 해석
# 878049 데이터가 있는데 실제 support를 보면 실제 데이터에 못미치기 때문에, f1-score값이 0이 나옴
```
