```python
import torch

# 1. 간단한 가중치(w) 초기화
w = torch.tensor([2.0], requires_grad=True)  # 초기값 2.0
print(f"초기 w: {w.item()}")

# 2. 간단한 손실함수 정의: L = (w - 5)^2
# 우리가 목표하는 값은 w = 5
def loss_fn(w):
    return (w - 5)**2

# 3. 옵티마이저 설정
optimizer = torch.optim.SGD([w], lr=0.1)  # 학습률 0.1

# 4. 1번 학습 스텝
optimizer.zero_grad()        # 기존에 남아있는 gradient 초기화
loss = loss_fn(w)            # 손실 계산
print(f"손실: {loss.item()}")
loss.backward()              # backward → 기울기 계산
print(f"계산된 기울기: {w.grad.item()}")

optimizer.step()             # optimizer → 가중치 업데이트
print(f"업데이트 후 w: {w.item()}")
```

```python
import random
import numpy as np

# 조기 종료 클래스
class EarlyStopping:
    def __init__(self, patience=3, min_delta=0):
        self.patience = patience
        self.min_delta = min_delta
        self.counter = 0
        self.best_loss = float('inf')
    
    def __call__(self, val_loss):
        if val_loss < self.best_loss - self.min_delta:
            self.best_loss = val_loss
            self.counter = 0
        else:
            self.counter += 1
            if self.counter >= self.patience:
                return True
        return False

# 객체 생성
early_stopping = EarlyStopping(patience=3)

# 가짜 검증 손실 값 (초기에는 줄다가 점점 변화없음음)
val_losses = [1.0, 0.8, 0.7, 0.7, 0.69, 0.69, 0.70, 0.71, 0.72]

# 시뮬레이션
    # enumerate(iterable, start=0)
    # iterable → 반복 가능한 객체 (리스트, 튜플, 문자열 등)
    # start → 인덱스 시작 값 (기본 0)
    # 반환값: (index, value) 튜플 순서
for epoch, val_loss in enumerate(val_losses, 1):  
    print(f"Epoch {epoch}, val_loss={val_loss:.4f}")
    
    if early_stopping(val_loss):
        print(f"→ 조기 종료 발생! (epoch={epoch})")
        break
```
