#### 🔹 머신러닝 모델을 개선하는 순서



#### 1️⃣ 데이터 준비
- 필요한 데이터 준비 → 결측치 제거, 이상치 확인
- 예시: 고객 정보, 구매 기록, 이탈 여부(Churn) 등



#### 2️⃣ 기본 모델 학습 (Baseline)
- 아무 튜닝 없이 기본 모델로 학습
- 예시: `LogisticRegression()`, `RandomForestClassifier()` 등
- 평가: accuracy, f1-score 등으로 기본 성능 확인

💡 목적:  
- 기준점(score) 만들기  
- “지금 모델이 어디까지 성능 내는지” 확인



#### 3️⃣ 학습 / 테스트 점수 비교
- Train vs Test 점수 확인
- Train 높고 Test 낮으면 → 과적합(overfitting)  
- 둘 다 낮으면 → 모델이 아직 잘 배우지 못함(underfitting)

💡 예시:  
- Train 0.95 / Test 0.65 → 과적합  
- Train 0.60 / Test 0.62 → underfitting



#### 4️⃣ 불균형 데이터 처리 (SMOTE 등)
- 클래스 비율이 심하게 차이나면 → 소수 클래스 놓칠 수 있음
- SMOTE, RandomOverSampler 등으로 데이터 균형 맞추기
- 이후 다시 모델 학습 → 성능 비교

💡 순서 팁:  
- 튜닝 전/후 둘 다 적용해보고 비교  
- 대부분은 SMOTE 적용 후 하이퍼파라미터 튜닝이 안정적



#### 5️⃣ 하이퍼파라미터 튜닝
- 모델 성능 개선 → `GridSearchCV`, `RandomizedSearchCV`
- 예시: `LogisticRegression`의 `C`, `solver`, `class_weight` 등
- 목적: Train/Test 점수 차이 줄이고 F1, Recall 등 원하는 지표 올리기

💡 팁:  
- CV(cross-validation) 사용 → 안정적인 점수 확인



#### 6️⃣ 최종 모델 선택
- 최종 모델 = `GridSearchCV.best_estimator_`
- Test set으로 최종 평가
- `classification_report`, `confusion_matrix` 확인

💡 중요 포인트:  
- 항상 Train/Test 비교 먼저 → 과적합/underfitting 확인  
- SMOTE 적용 → 튜닝 순서가 흔함  
- 최종 평가는 `GridSearchCV.best_estimator_` 사용

#### 🔹 데이터 불균형 해결방법

1. 모델 기반 접근: 클래스 가중치(class_weight)

- 다수 클래스보다 소수 클래스에 더 큰 **중요도(weight)**를 줘서 학습
- 장점: 간단, 데이터 증강 필요 없음
- 단점: 데이터가 매우 불균형하면 성능 제한 가능
- 사용 예시: LogisticRegression, RandomForest, SVM 등 대부분의 sklearn 분류기에서 지원
<br> clf = RandomForestClassifier(class_weight='balanced')
- 직접계산
<br> from sklearn.utils.class_weight import compute_class_weight
<br> weights = compute_class_weight('balanced', classes=[0,1], y=y_train)
<br> clf = LogisticRegression(class_weight={0: weights[0], 1: weights[1]})


2. 샘플링 기반 접근: 언더샘플링 / 오버샘플링
- SMOTE 말고도 단순 샘플링 방법 있음
    1. 언더샘플링 (Under-sampling)
        - 다수 클래스의 데이터를 일부만 사용해서 균형 맞춤
        - 장점: 학습 속도 빠름, 데이터 많을 때 유용
        - 단점: 정보 손실 가능
        - sklearn: RandomUnderSampler
<br>        from imblearn.under_sampling import RandomUnderSampler
<br>        rus = RandomUnderSampler(random_state=42)
<br>        X_res, y_res = rus.fit_resample(X_train, y_train)
    2. 오버샘플링 (Over-sampling)
        - 소수 클래스 데이터를 반복해서 늘려 균형 맞춤
        - 장점: 데이터 손실 없음
        - 단점: 단순 복제라 과적합 가능
        - sklearn: RandomOverSampler
<br>        from imblearn.over_sampling import RandomOverSampler
<br>        ros = RandomOverSampler(random_state=42)
<br>        X_res, y_res = ros.fit_resample(X_train, y_train)

    3. 오버샘플링 SMOTE (Synthetic Minority Over-sampling Technique)
        - 소수 클래스 데이터를 주변 샘플 간 선형 보간으로 합성
        - 장점: 단순 복제보다 과적합 덜 발생
        - 단점: 합성 데이터가 실제 분포와 완전히 일치하지 않을 수 있음
        - sklearn: SMOTE
    <br>        from imblearn.over_sampling import SMOTE
    <br>        smote = SMOTE(random_state=42)
    <br>        X_res, y_res = smote.fit_resample(X_train, y_train)

    4. 오버샘플링SMOTEENN (SMOTE + Edited Nearest Neighbors)
        - SMOTE로 소수 클래스 증강 후 ENN 적용 → 노이즈 제거
        - 장점: 소수 클래스 증강 + 데이터 정제 가능
        - 단점: 학습 데이터 변형이 복잡하고, 학습 시간이 증가할 수 있음
        - sklearn: SMOTEENN
    <br>        from imblearn.combine import SMOTEENN
    <br>        smote_enn = SMOTEENN(random_state=42)
    <br>        X_res, y_res = smote_enn.fit_resample(X_train, y_train)

3. 앙상블 기반 접근
- 다수 클래스에 치우친 데이터에서도 균형 잡힌 성능을 낼 수 있음
- 대표적인 방법:
    - Balanced Random Forest
        - RandomForest 기반, 각 트리를 학습할 때 언더샘플링 적용
<br>        from imblearn.ensemble import BalancedRandomForestClassifier
<br>        clf = BalancedRandomForestClassifier(random_state=42)
    - EasyEnsemble
        - 여러 개의 언더샘플링된 데이터셋으로 모델 학습 후 앙상블
<br>        from imblearn.ensemble import EasyEnsembleClassifier
<br>        clf = EasyEnsembleClassifier(random_state=42)

#### 🔹 SMOTE, SMOTEENN

- 데이터가 많고 깨끗하면 → SMOTE만
- 데이터가 noisy(헷갈리는 샘플 많음)하면 → SMOTEENN


1. SMOTE (소수 클래스 늘리기)
- 상황 예시:
<br>    반 친구들 중에서 남자 2명, 여자 8명이 있어요.
<br>    남자 친구 데이터를 더 늘리고 싶은데, 그냥 똑같이 복사하면 똑같은 친구가 계속 생기는 것처럼 모델이 헷갈려요.

- SMOTE 방식:
<br>    남자 친구 2명 사이를 이어서 새로운 가상의 남자 친구를 만들어내요.
<br>    예를 들어, 두 남자 친구 키가 160cm, 170cm면, 165cm 정도인 새 친구를 만들어서 남자 데이터를 늘리는 느낌이에요.

- 정리:
    - 그냥 복사 X
    - 두 샘플 사이에 새로운 가짜 데이터 생성
    - 데이터가 다양해져서 모델이 덜 헷갈림


2. SMOTEENN (SMOTE + 노이즈 제거)
- 상황 예시:
<br>    SMOTE로 남자 친구를 늘렸는데, 주변에 여자 친구와 헷갈릴 수 있는 애들이 섞여 있어요.
<br>    예를 들어, 키 160cm인데 옷 스타일이 여자 친구랑 비슷한 애가 있을 수 있어요.

- SMOTEENN 방식:
<br>   먼저 SMOTE처럼 가짜 남자 친구를 만들어요.
<br>   그 다음, 주변 친구들을 확인해서 **잘못 섞인 친구(노이즈)**를 삭제해요.
<br>   즉, 남자 친구 데이터가 깨끗해지는 것

- 정리:
    - SMOTE로 늘린 후 주변 헷갈리는 애들 제거
    - 더 깨끗하고 모델이 잘 배울 수 있는 데이터가 됨

3. 한 줄 요약
- SMOTE: 남자 친구 수를 늘림 → 다양하지만 조금 헷갈릴 수 있음
- SMOTEENN: 남자 친구 수 늘리고 → 주변 헷갈리는 애들은 삭제 → 더 깨끗함

#### PCA (Principal Component Analysis, 주성분 분석)
1. PCA란?
- 데이터가 10개의 특성을 가지고 있는데
- → 사실 그 중 2~3개 특성만으로도 데이터의 특징을 잘 설명할 수 있다면
- → PCA를 이용해서 2~3개 차원으로 줄여서 분석, 시각화, 학습을 쉽게 만들 수 있어요.

2. 왜 쓰는가?
- 차원 축소 → 연산이 빨라짐, 메모리 적게 사용
- 노이즈 제거 → 중요하지 않은 정보는 버리고 핵심 정보만 남김
- 시각화 → 2D, 3D로 데이터 시각화 가능
<br> 예: 100개의 특성을 가진 데이터 → PCA로 2차원으로 줄여서 산점도로 보기

3. 핵심 아이디어
- 데이터의 분산이 가장 큰 방향을 찾아서 그 방향을 기준으로 축을 새로 만듦
- 이 새 축을 **주성분(principal component)**이라고 부름
- 1번째 주성분: 가장 큰 분산 방향 ---> 중요한 정보가 담겨있음.
- 2번째 주성분: 1번째 주성분과 직각(서로 독립)인 방향으로 가장 큰 분산

| **군집 / 비지도 학습** | K-Means            | 데이터 안에 비슷한 그룹이 있을 때               | 군집 수(k) 직접 설정, 원형/균등 분포 적합        |
| **군집 / 비지도 학습** | Hierarchical       | 데이터 간 계층 구조가 필요할 때                | 덴드로그램으로 군집 구조 시각화, 클러스터 개수 유연     |
| **군집 / 비지도 학습** | DBSCAN             | 노이즈가 많고 군집 모양이 불규칙할 때             | 밀도 기반 군집화, 이상치(Noise) 자동 구분       |

### 🔹 DBSCAN (비지도 학습)
🧠 특징 요약
DBSCAN은 "가까이 몰려 있는 애들끼리 묶고, 혼자 있는 애는 버리는" 군집화 방법이에요.

💬 그룹 수는 자동으로 정해지고, 이상치도 알아서 걸러줘요!

- K-Means처럼 k를 지정하지 않아도 됨
